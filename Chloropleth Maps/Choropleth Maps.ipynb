```python
import pandas as pd
import numpy as np
import geopandas as gpd
import plotly.express as px
import matplotlib.pyplot as plt
```

### Current Residents

```python
current_data = pd.read_excel(r'D:\RA\GitHub\Immune-Disparities\Chloropleth Maps\choropleth_data.xlsx', sheet_name="Current Census Region")
```

```python
# Choropleth map of USA
fig = px.choropleth(current_data[current_data["immune cell"] == "Neutrophils"], locations='code',
                    locationmode="USA-states", color='counts', scope="usa",
                   hover_name="region",
                   color_continuous_scale=px.colors.sequential.RdBu)

fig.update_layout(
    title={
        'text': "<b>Neutrophils</b>",  
        'x': 0.5,  
        'xanchor': 'center',
        'yanchor': 'top',
        'font': {'size': 24}
    },
    coloraxis_colorbar={
        'yanchor': 'middle',
        'x': 0.70
    }
)
 
fig.show()
```
